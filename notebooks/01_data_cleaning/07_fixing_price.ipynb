## 🧠  Party Classification Model

**Goal:**  
Train a machine learning model to predict the `party` column for rows where it is currently missing, based on the available features such as:

- `headline` (text)
- `price`, `price_basis`, `location`, etc. (structured features)

**Why this matters:**  
Around 10,000+ rows have missing `party` values. Many of these are repetitive benchmarks, but some may be classifiable based on pricing behavior and text patterns. By using a neural network, we aim to:

- Learn non-obvious patterns in text and pricing,
- Predict party labels for incomplete entries,
- Improve dataset completeness for downstream analysis.

We will:
1. Preprocess the dataset,
2. Tokenize and vectorize the `headline`,
3. Combine it with structured features,
4. Train a neural network to predict the party,
5. Apply the model to fill missing values where confidence is high.


```python
# Print % filled and total rows/columns
filled_percent = (main.notnull().mean() * 100).round(2)
print("📊 % of Filled Values per Column:\n")
print(filled_percent)

print("\n📄 Total Rows:", len(main))
print("📈 Total Columns:", len(main.columns))
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.feature_extraction.text import TfidfVectorizer
import numpy as np

# 1. Encode party labels (only non-null)
main_non_null = main[main['party'].notnull()].copy()
label_encoder = LabelEncoder()
main_non_null['party_encoded'] = label_encoder.fit_transform(main_non_null['party'])

# 2. Vectorize headline text with TF-IDF (limit features to 5000 for speed)
tfidf = TfidfVectorizer(max_features=5000, stop_words='english')
X_text = tfidf.fit_transform(main_non_null['headline'])

# 3. Prepare numeric features - price, price_basis (one-hot), volume_min/max, location (one-hot)
# One-hot encode categorical columns
price_basis_ohe = pd.get_dummies(main_non_null['price_basis'], prefix='price_basis')
location_ohe = pd.get_dummies(main_non_null['location'], prefix='loc')

# Combine numeric features
numeric_feats = main_non_null[['price', 'volume_min', 'volume_max']].fillna(0)
scaler = StandardScaler()
numeric_scaled = scaler.fit_transform(numeric_feats)

# Combine all features horizontally
from scipy.sparse import hstack
X = hstack([X_text, price_basis_ohe.values, location_ohe.values, numeric_scaled])

# Labels
y = main_non_null['party_encoded'].values

# 4. Split into train and test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.15, random_state=42, stratify=y)

print(f"Training samples: {X_train.shape[0]}, Testing samples: {X_test.shape[0]}")
```

```python
import numpy as np
from collections import Counter

# Count samples per class
class_counts = Counter(main_non_null['party_encoded'])
valid_classes = [cls for cls, count in class_counts.items() if count > 1]

# Filter dataset to only include valid classes
mask_valid = main_non_null['party_encoded'].isin(valid_classes)
main_filtered = main_non_null[mask_valid]

# Re-extract features and labels
y_filtered = main_filtered['party_encoded'].values
X_text_filtered = tfidf.transform(main_filtered['headline'])

price_basis_ohe_filtered = pd.get_dummies(main_filtered['price_basis'], prefix='price_basis')
location_ohe_filtered = pd.get_dummies(main_filtered['location'], prefix='loc')
numeric_feats_filtered = main_filtered[['price', 'volume_min', 'volume_max']].fillna(0)
numeric_scaled_filtered = scaler.transform(numeric_feats_filtered)

from scipy.sparse import hstack
X_filtered = hstack([X_text_filtered, price_basis_ohe_filtered.values, location_ohe_filtered.values, numeric_scaled_filtered])

# Train-test split with filtered data
X_train, X_test, y_train, y_test = train_test_split(
    X_filtered, y_filtered, test_size=0.15, random_state=42, stratify=y_filtered)

print(f"Training samples: {X_train.shape[0]}, Testing samples: {X_test.shape[0]}")
```

```python
import torch
from torch.utils.data import Dataset, DataLoader

class PartyDataset(Dataset):
    def __init__(self, X_sparse, y):
        self.X = X_sparse
        self.y = y

    def __len__(self):
        return self.X.shape[0]

    def __getitem__(self, idx):
        # Convert sparse row to dense tensor
        x = torch.tensor(self.X[idx].toarray(), dtype=torch.float32).squeeze(0)
        y = torch.tensor(self.y[idx], dtype=torch.long)
        return x, y

# Create datasets
train_dataset = PartyDataset(X_train, y_train)
test_dataset = PartyDataset(X_test, y_test)

# Create data loaders
batch_size = 256
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False)
```

```python
import torch.nn as nn
import torch.nn.functional as F

class PartyClassifier(nn.Module):
    def __init__(self, input_dim, num_classes):
        super(PartyClassifier, self).__init__()
        self.fc1 = nn.Linear(input_dim, 1024)
        self.dropout1 = nn.Dropout(0.3)
        self.fc2 = nn.Linear(1024, 512)
        self.dropout2 = nn.Dropout(0.3)
        self.fc3 = nn.Linear(512, num_classes)

    def forward(self, x):
        x = F.relu(self.fc1(x))
        x = self.dropout1(x)
        x = F.relu(self.fc2(x))
        x = self.dropout2(x)
        x = self.fc3(x)
        return x

input_dim = X_train.shape[1]
num_classes = len(np.unique(y_train))

model = PartyClassifier(input_dim, num_classes).cuda()
```

```python
import torch.optim as optim

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=1e-3)

def train_epoch(model, loader, optimizer, criterion):
    model.train()
    total_loss = 0
    for x_batch, y_batch in loader:
        x_batch, y_batch = x_batch.cuda(), y_batch.cuda()
        optimizer.zero_grad()
        outputs = model(x_batch)
        loss = criterion(outputs, y_batch)
        loss.backward()
        optimizer.step()
        total_loss += loss.item() * x_batch.size(0)
    return total_loss / len(loader.dataset)
```

```python
def validate_epoch(model, loader, criterion):
    model.eval()
    total_loss = 0
    correct = 0
    with torch.no_grad():
        for x_batch, y_batch in loader:
            x_batch, y_batch = x_batch.cuda(), y_batch.cuda()
            outputs = model(x_batch)
            loss = criterion(outputs, y_batch)
            total_loss += loss.item() * x_batch.size(0)
            preds = outputs.argmax(dim=1)
            correct += (preds == y_batch).sum().item()
    accuracy = correct / len(loader.dataset)
    return total_loss / len(loader.dataset), accuracy
```

```python
num_epochs = 5
best_val_acc = 0

for epoch in range(1, num_epochs + 1):
    train_loss = train_epoch(model, train_loader, optimizer, criterion)
    val_loss, val_acc = validate_epoch(model, test_loader, criterion)
    print(f"Epoch {epoch}: Train Loss={train_loss:.4f}, Val Loss={val_loss:.4f}, Val Acc={val_acc:.4f}")

    # Save best model
    if val_acc > best_val_acc:
        best_val_acc = val_acc
        torch.save(model.state_dict(), "best_party_classifier.pth")
        print("Best model saved.")
```

```python
import torch
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
import numpy as np
import pandas as pd

# Custom Dataset handling sparse matrix indexing properly
class NullPartyDataset(Dataset):
    def __init__(self, X):
        # Convert to CSR format for efficient row slicing
        self.X = X.tocsr()

    def __len__(self):
        return self.X.shape[0]

    def __getitem__(self, idx):
        x = torch.tensor(self.X[idx].toarray(), dtype=torch.float32).squeeze(0)
        return x

# Prepare data for null-party rows
null_party_df = main[main['party'].isnull()].copy()

# Vectorize headline text
X_text_null = tfidf.transform(null_party_df['headline'])

# One-hot encode price_basis and location, align columns with training data
price_basis_ohe_null = pd.get_dummies(null_party_df['price_basis'], prefix='price_basis')
location_ohe_null = pd.get_dummies(null_party_df['location'], prefix='loc')

# Add missing columns with zeros
for col in price_basis_ohe.columns:
    if col not in price_basis_ohe_null:
        price_basis_ohe_null[col] = 0
price_basis_ohe_null = price_basis_ohe_null[price_basis_ohe.columns]

for col in location_ohe.columns:
    if col not in location_ohe_null:
        location_ohe_null[col] = 0
location_ohe_null = location_ohe_null[location_ohe.columns]

# Scale numeric features
numeric_null = null_party_df[['price', 'volume_min', 'volume_max']].fillna(0)
numeric_null_scaled = scaler.transform(numeric_null)

from scipy.sparse import hstack
X_null = hstack([X_text_null, price_basis_ohe_null.values, location_ohe_null.values, numeric_null_scaled])

# Create Dataset and DataLoader
null_dataset = NullPartyDataset(X_null)
null_loader = DataLoader(null_dataset, batch_size=256, shuffle=False)

# Load best model and set to eval
model.load_state_dict(torch.load("best_party_classifier.pth"))
model.eval()

# Predict with confidence thresholding
predicted_parties = []
confidences = []

with torch.no_grad():
    for batch in null_loader:
        batch = batch.cuda()
        outputs = model(batch)
        probs = F.softmax(outputs, dim=1)
        conf, preds = torch.max(probs, dim=1)
        predicted_parties.extend(preds.cpu().numpy())
        confidences.extend(conf.cpu().numpy())

# Apply confidence threshold (e.g., 0.9)
threshold = 0.9
final_predictions = []
for conf, pred in zip(confidences, predicted_parties):
    if conf >= threshold:
        final_predictions.append(label_encoder.inverse_transform([pred])[0])
    else:
        final_predictions.append(None)

# Update main DataFrame
main.loc[null_party_df.index, 'party'] = final_predictions
```

This step improves:

- Accuracy of party-level analyses,
- Performance and reliability of machine learning models,
- Clarity in market interpretation.

This decision is consistent with best practices in commodity trade data preprocessing.
